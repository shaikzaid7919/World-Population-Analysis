```python
df.rename(columns={'Country/Territory':'Country'},inplace=True)
```

```python
top10 = df.sort_values('2022 Population', ascending=False).head(10)
plt.figure(figsize=(10,6))
sns.barplot(data=top10, x='Country', y='2022 Population')
plt.title('Top 10 Most Populated Countries (2022)', fontsize=16)
plt.xticks(rotation=45)
plt.ylabel('Population')
plt.xlabel('Country')
plt.tight_layout()

```

```python
continent_pop = df.groupby('Continent')['2020 Population'].sum()
plt.figure(figsize=(6,6))
colors = sns.color_palette('pastel')
plt.pie(continent_pop, labels=continent_pop.index, autopct='%1.1f%%', colors=colors, startangle=140)
plt.title('World Population Share by Continent (2020)', fontsize=16)
plt.show()
```

```python
countries = ['India', 'China', 'United States']
plt.figure(figsize=(12,6))
for country in countries:
    subset = df[df['Country'] == country]
    years = ['1970 Population', '1980 Population', '1990 Population', '2000 Population', '2010 Population', '2015 Population', '2020 Population', '2022 Population']
    populations = subset[years].values.flatten()
    plt.plot(years, populations, marker='o', label=country)

plt.title('Population Growth Over Time (1970–2022)', fontsize=16)
plt.xlabel('Year')
plt.ylabel('Population')
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
corr_data = df[['Area (km²)', 'Density (per km²)', 'Growth Rate', '2022 Population']]
plt.figure(figsize=(8,6))
sns.heatmap(corr_data.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Area (km²)', y='2022 Population', hue='Continent', size='Density (per km²)', palette='muted', sizes=(20,200))
plt.xscale('log')
plt.yscale('log')
plt.title('Area vs Population (Colored by Continent)', fontsize=16)
plt.xlabel('Area (log scale)')
plt.ylabel('Population (log scale)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
fig = px.choropleth(df,
                    locations='CCA3',
                    color='Density (per km²)',
                    hover_name='Country',
                    color_continuous_scale='YlOrRd',
                    title='World Population Density by Country',width=1000,height=500)

fig.update_layout(geo=dict(showframe=False, showcoastlines=True))
fig.show()
```
